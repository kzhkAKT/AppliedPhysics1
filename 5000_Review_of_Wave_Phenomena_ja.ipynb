## 2. 幾何光学とスネルの法則

光が二つの透明な物質の境界面に達すると、一部は反射し、一部は進む向きを変えて透過します。この進む向きの変化を**屈折**といいます。各物質には**屈折率** $n$ が定まっており、真空中の光速と物質中の光速の比 $n = c / v$ で定義されます。

```python
#@title Geometrical optics and Snell's law / 幾何光学とスネルの法則
%%html
<iframe src="https://phet.colorado.edu/sims/html/bending-light/latest/bending-light_ja.html"
        width="800"
        height="600"
        allowfullscreen>
</iframe>
```

**スネルの法則**

\begin{align*}
n_1 \sin \theta_1 &= n_2 \sin \theta_2 .
\end{align*}

![](https://upload.wikimedia.org/wikipedia/commons/thumb/3/3f/Snells_law2.svg/267px-Snells_law2.svg.png)

図では、光線が媒質 1（屈折率 $n_1$）の点 P から境界面上の点 O へ進み、さらに媒質 2（屈折率 $n_2$）の点 Q へ進んでいます。角 $\theta_1$, $\theta_2$ は**法線**（境界面に垂直な直線）から測ります。

- $n_1$ に対する $n_2$ の相対屈折率 $n_{12}$（媒質 1 を基準としたときの媒質 2 の屈折率）を、$n_1$, $n_2$ を用いて表せ。
- 光線を逆向きに Q → O → P と進ませ、媒質 2 から媒質 1 へ入射させる。入射角が十分大きいと、光は媒質 1 へ進まずすべて反射される（**全反射**）。屈折光が境界面をかすめる（$\theta_1 = 90^\circ$ となる）ときの入射角を**臨界角** $\theta_c$ という。$\theta_c$ を求めよ。

## 8. 定常波

同じ振幅・同じ波長の二つの波が同一直線上を逆向きに進むとき、その重ね合わせはどこへも進まない波になります。これが**定常波**です。まったく振動しない点を**節**、節と節の中間で最も大きく振動する点を**腹**といいます。対話型グラフ、実験動画、GeoGebra アプレットで確かめましょう。

## 9. 波の干渉と回折

二つの波源からの波が重なると、強め合う点と打ち消し合う点ができます。この模様を**干渉**といいます。また、波はすき間を通ったり障害物の縁を回り込んだりして広がります。これが**回折**で、すき間の幅が波長と同程度のときに最も顕著になります。

```python
#@title Interference and diffraction of waves / 波の干渉と回折
%%html
<iframe src="https://phet.colorado.edu/sims/html/wave-interference/latest/wave-interference_ja.html"
        width="800"
        height="600"
        allowfullscreen>
</iframe>
```

- 二つの点波源が、同じ振動数で同位相に振動している。二つの波源から点 P までの距離を $r_1$, $r_2$ とするとき、P が強く振動する点（強め合う干渉）となる条件と、弱く振動する点（弱め合う干渉）となる条件を、経路差 $|r_1 - r_2|$ について導け。
