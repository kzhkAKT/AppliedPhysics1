```
#@title 振動数の測定
%%html
<iframe width="560" height="315" src="https://www.youtube-nocookie.com/embed/02w9lSii_Hs?si=qF7rfpGv5SjiwS8Y" title="YouTube video player" frameborder="0" allow="accelerometer; autoplay; clipboard-write; encrypted-media; gyroscope; picture-in-picture; web-share" referrerpolicy="strict-origin-when-cross-origin" allowfullscreen></iframe>
```

教科書 上巻 p76 図1.103

$\theta << 1$
\begin{align*}
&\begin{cases}
m a &= - mg \sin \theta, \quad \text{equation of motion}\\
\sin \theta &= \dfrac{x}{L}, \quad \text{geometrical relation}
\end{cases}\\
m a_x &= - m\dfrac{g}{L}x\\
 a_x &= - \dfrac{g}{L}x\\
 \dfrac{{\rm d^2} x}{{\rm d}t^2} &= - \dfrac{g}{L}x\\
 &x = A \sin \left( \omega  t + \delta \right), \quad \omega^2 = \dfrac{g}{L}\\
\end{align*}

この関係を利用し, 角振動数/角速度 $\omega$ を整理すると
\begin{align*}
\omega &= \dfrac{2\pi}{T} = 2\pi f\\
f &= \dfrac{\omega}{2\pi} = \dfrac{1}{2\pi} \sqrt{\dfrac{g}{L}}
\end{align*}

このことから振動数 $f$ は $f(L)$ と $L$ の関数になることが分かります

周期 $T$ は
$$
T = 2\pi\sqrt{\dfrac{L}{g}}
$$
と表されます.
