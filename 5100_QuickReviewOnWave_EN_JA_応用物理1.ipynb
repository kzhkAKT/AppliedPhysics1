> 💡 A **wave** is a vibration passed from one place to the next through a **medium** (rope, water, air…). Each part of the medium only moves back and forth; what travels is the **shape and the energy**. ｜ **波**とは、振動が**媒質**（ロープ・水・空気…）を通して次々に伝わる現象。媒質はその場で振動するだけで、進むのは**波形とエネルギー**。

## 1. Waves on a string ｜ 直線上を伝わる波 　*(PhET: Wave on a String)*

$$ fT = 1, \qquad v = \frac{\lambda}{T} = f\lambda, \qquad y(x,t) = A\sin 2\pi\!\left(\frac{t}{T} - \frac{x}{\lambda}\right) $$

$A$ amplitude 振幅, $\lambda$ wavelength 波長, $T$ period 周期, $f$ frequency 振動数, $v$ wave speed 波の速さ.
The **source** sets $f$; the **medium** sets $v$. ｜ $f$ は**波源**が、$v$ は**媒質**が決める。

**Try it ｜ やってみよう** — the settings are the buttons inside the simulation ｜ 設定はシミュレーション内のボタン

1. **Pulse · No End · Slow Motion · Damping: None** → send one pulse. The green beads only move up and down; the pulse moves to the right. ｜ パルスを 1 つ送る。緑のビーズは上下に動くだけで、進むのはパルス。
2. **Oscillate · No End** → change **Amplitude** and **Frequency**; measure $\lambda$ with the **Ruler**. Higher $f$ → shorter $\lambda$ (the string fixes $v$). ｜ 振幅・振動数を変え、定規で $\lambda$ を測る。$f$ を上げると $\lambda$ は短くなる（$v$ はひもで決まる）。
3. **Pulse · Fixed End**, then **Pulse · Loose End** → fixed end: the crest returns as a trough (phase shift $\pi$). Loose (free) end: the crest returns as a crest. ｜ 固定端：山が谷になって返る（位相が $\pi$ ずれる）。自由端：山のまま返る。
4. **Oscillate · Fixed End · Damping: None** → tune the frequency until you get a clean **standing wave**. The **nodes** never move; node spacing $= \lambda/2$. ｜ 振動数を調整してきれいな**定常波**を作る。**節**は動かない。節の間隔は $\lambda/2$。

## 2. Longitudinal waves & wavefronts ｜ 縦波と波面 　*(PhET: Waves Intro)*

- **Transverse** 横波: the medium moves **perpendicular** to the direction of travel (rope, light, S-wave). ｜ 媒質の振動が進行方向と**垂直**（ロープ、光、S 波）。
- **Longitudinal** 縦波: the medium moves **parallel** to the direction of travel, making dense and sparse regions (sound, P-wave). ｜ 媒質の振動が進行方向と**平行**。密と疎が交互にできる＝疎密波（音、P 波）。
- **Wavefront** 波面: a line through points with the same phase (all the crests). **Rays** 射線 are perpendicular to it. **Huygens' principle**: every point on a wavefront sends out a small wavelet; the envelope of the wavelets is the next wavefront. ｜ 同じ位相の点を結んだ線。射線は波面に垂直。**ホイヘンスの原理**：波面上の各点から出る素元波の包絡面が次の波面。

**Try it ｜ やってみよう**

1. **Sound** screen → view **Particles** (or **Both**): the air molecules bunch up and spread out. ｜ Sound 画面で Particles（または Both）：空気の粒が詰まったり広がったりする。
2. **Water** screen → the rings are **wavefronts**. Raise the frequency: the rings get closer ($\lambda$ shorter). ｜ Water 画面の輪が**波面**。振動数を上げると輪の間隔（$\lambda$）が縮む。

❓ Earthquakes send P-waves and S-waves. Which one is longitudinal? Which arrives first? ｜ 地震の P 波と S 波、縦波はどちら？ 先に届くのは？

## 3. Interference & diffraction ｜ 干渉と回折 　*(PhET: Wave Interference)*

Two sources in phase; path difference $\Delta L = \lvert\, \mathrm{S_1P} - \mathrm{S_2P} \,\rvert$ ｜ 同位相の 2 つの波源、経路差 $\Delta L$：

$$ \text{strong 強め合う: } \Delta L = m\lambda \qquad\qquad \text{weak 弱め合う: } \Delta L = \left(m + \tfrac12\right)\lambda \qquad (m = 0, 1, 2, \dots) $$

- The calm (weak) lines are **hyperbolas** with $\mathrm{S_1}$, $\mathrm{S_2}$ as foci. ｜ 静かな線は $\mathrm{S_1}$, $\mathrm{S_2}$ を焦点とする**双曲線**。
- **Diffraction** 回折: waves spread out after a gap or around an edge — most strongly when the gap width $d \approx \lambda$. ｜ すき間や端で波が回り込む。すき間の幅 $d \approx \lambda$ のとき最も大きく広がる。

**Try it ｜ やってみよう**

1. **Interference** screen → turn on both sources → find the lines where the water stays calm. Change **Separation**: more lines or fewer? ｜ Interference 画面：波源を 2 つとも ON。水面が静かな線を探す。Separation を変えると線の数は？
2. **Slits** screen → **One Slit** → make the slit narrow (≈ $\lambda$): the wave spreads out like a circle. Make it wide: it goes almost straight through. ｜ Slits 画面で One Slit：幅を $\lambda$ 程度に → 円形に広がる。広げる → ほぼまっすぐ進む。

## 4. Reflection, refraction, total internal reflection ｜ 反射・屈折・全反射 　*(PhET: Bending Light)*

$$ \theta_i = \theta_r \qquad\qquad n_1\sin\theta_1 = n_2\sin\theta_2,\quad \frac{n_2}{n_1} = \frac{\sin\theta_1}{\sin\theta_2} = \frac{v_1}{v_2} = \frac{\lambda_1}{\lambda_2} \qquad\qquad \sin\theta_c = \frac{n_2}{n_1} $$

- Angles are measured from the **normal**. At the boundary $f$ stays the same; $v$ and $\lambda$ change. ｜ 角度は**法線**から測る。境界で $f$ は変わらず、$v$ と $\lambda$ が変わる。
- Into a **slower** medium (larger $n$): the ray bends **toward** the normal and $\lambda$ gets shorter. ｜ **遅い**媒質（$n$ が大きい）へ：射線は法線に**近づき**、$\lambda$ は短くなる。
- **Total internal reflection** 全反射 happens only going into a **faster** medium ($n_1 > n_2$) with $\theta_1 > \theta_c$ — optical fibres, diamonds. ｜ **速い**媒質（$n_1 > n_2$）へ進むときだけ、$\theta_1 > \theta_c$ で起こる。光ファイバー、ダイヤモンド。

**Try it ｜ やってみよう**

1. **Intro** screen → switch the laser on and rotate it. Drag out the **protractor**: check $\theta_r = \theta_i$ and read $\theta_2$. ｜ Intro 画面：レーザーを ON にして回す。分度器で反射角＝入射角を確かめ、屈折角を読む。
2. Switch to the **Wave** view: in glass the wavefronts are closer together ($\lambda$ shorter, $v$ slower). ｜ Wave 表示に切り替え：ガラスの中は波面の間隔が狭い（$\lambda$ 短い、$v$ 遅い）。
3. Set **top = Glass, bottom = Air** and increase the angle slowly: the refracted beam disappears at $\theta_c \approx 42°$. ｜ 上をガラス、下を空気にして角度を少しずつ大きく：約 42° で屈折光が消える。

```python
#@title ▶ PhET: Bending Light ｜ 光の屈折（反射・屈折・全反射）
from IPython.display import IFrame, display
display(IFrame("https://phet.colorado.edu/sims/html/bending-light/latest/bending-light_en.html", width=800, height=500))
```

## 📌 Summary ｜ まとめ

| Formula / rule ｜ 式・ルール | |
|---|---|
| $fT = 1,\quad v = \lambda/T = f\lambda$ | basic equation ｜ 波の基本式 |
| $y = A\sin 2\pi\left(\dfrac{t}{T} - \dfrac{x}{\lambda}\right)$ | sine wave travelling in $+x$ ｜ $+x$ 向きの正弦波 |
| fixed end: phase $+\pi$ (crest → trough) ・ free end: no change | reflection at an end ｜ 固定端・自由端 |
| $y = 2A\sin kx\,\cos\omega t$, nodes every $\lambda/2$ | standing wave ｜ 定常波 |
| strong $\Delta L = m\lambda$ ・ weak $\Delta L = (m + \tfrac12)\lambda$ | interference ｜ 干渉 |
| strongest when $d \approx \lambda$ | diffraction ｜ 回折 |
| $\theta_i = \theta_r$ ・ $n_1\sin\theta_1 = n_2\sin\theta_2$ ・ $\sin\theta_c = n_2/n_1$ (needs $v_1 < v_2$) | reflection ・ refraction ・ TIR ｜ 反射・屈折・全反射 |
