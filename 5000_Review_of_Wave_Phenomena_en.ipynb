## 2. Geometrical optics and Snell's law

When light reaches the boundary between two transparent materials, part of it is reflected and part passes through with a change of direction. The change of direction is called *refraction*. Each material has a *refractive index* $n$, the ratio of the speed of light in vacuum to its speed in the material: $n = c / v$.

```python
#@title Geometrical optics and Snell's law / 幾何光学とスネルの法則
%%html
<iframe src="https://phet.colorado.edu/sims/html/bending-light/latest/bending-light_en.html"
        width="800"
        height="600"
        allowfullscreen>
</iframe>
```

**Snell's law**

\begin{align*}
n_1 \sin \theta_1 &= n_2 \sin \theta_2 .
\end{align*}

![](https://upload.wikimedia.org/wikipedia/commons/thumb/3/3f/Snells_law2.svg/267px-Snells_law2.svg.png)

In the figure, a ray travels from P in medium 1 (refractive index $n_1$) to the point O on the boundary, and continues to Q in medium 2 (refractive index $n_2$). The angles $\theta_1$ and $\theta_2$ are measured from the *normal*, the line perpendicular to the boundary.

- Find the *relative* refractive index $n_{12}$ of medium 2 with respect to medium 1 (that is, the refractive index of medium 2 when medium 1 is taken as the reference) in terms of $n_1$ and $n_2$.
- Now reverse the ray so that it travels Q → O → P, from medium 2 into medium 1. If the angle of incidence is large enough, no light passes into medium 1 and all of it is reflected: *total internal reflection*. The angle of incidence at which the refracted ray just grazes the boundary ($\theta_1 = 90^\circ$) is the *critical angle* $\theta_c$. Find $\theta_c$.

## 3. Lenses

## 5. Transverse and longitudinal waves

Waves are classified by the direction in which the medium vibrates compared with the direction in which the wave travels. In a *transverse* wave the medium vibrates perpendicular to the direction of travel (first video). In a *longitudinal* wave the medium vibrates back and forth along the direction of travel, producing regions of compression and rarefaction (second video).

## 8. Standing waves

When two waves with the same amplitude and wavelength travel in opposite directions along the same line, their sum does not travel anywhere: it is a *standing wave*. Some points, the *nodes*, never move; halfway between them, the *antinodes* vibrate with the largest amplitude. ## 9. Interference and diffraction

When waves from two sources overlap, they reinforce each other at some points and cancel each other at others. This pattern is called *interference*. Waves also spread out when they pass through a gap or around the edge of an obstacle; this is *diffraction*, and it is most noticeable when the gap is about the same size as the wavelength.

## 10. Huygens' principle

Every point on a wavefront can be regarded as the source of a small circular (or spherical) wave. The new wavefront a moment later is the surface that touches all of these small waves at once (their *envelope*). This simple construction explains how waves propagate, reflect and refract.
